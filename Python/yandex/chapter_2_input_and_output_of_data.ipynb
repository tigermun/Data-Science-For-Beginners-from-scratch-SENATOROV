```python
# 12. 

n, m = input(), input()
max_length = max(len(n), len(m))
n = n.zfill(max_length)
m = m.zfill(max_length)
result: str = ""
for i in range(max_length):
    digit = (int(n[i]) + int(m[i])) % 10
    result += str(digit)
print(int(result))
```
