```python
# 4.  

class Programmer:

    def __init__(self, name, level):
        self.name = name
        self.level = level
        self.hours = 0
        self.salary = {"Junior": 10, "Middle": 15, "Senior": 20}[self.level]
        self.money = 0

    def work(self, time):
        self.money = self.money + self.salary * time
        self.hours += time

    def rise(self):
        if self.level == "Senior":
            self.salary += 1
        elif self.level == "Junior":
            self.level = "Middle"
            self.salary = 15
        else:
            self.level = "Senior"
            self.salary = 20

    def info(self):
        return f"{self.name} {self.hours}ч. {self.money}тгр."
```

```python
# 7.  

class Rectangle:

    def __init__(self, corner1, corner2):
        self.corner1 = corner1
        self.corner2 = corner2
        self.length = abs(corner2[0] - corner1[0])
        self.width = abs(corner2[1] - corner1[1])

    def perimeter(self):
        return round(2 * (self.length + self.width), 2)

    def area(self):
        return round(self.length * self.width, 2)

    def get_pos(self):
        x1, y1 = self.corner1
        x2, y2 = self.corner2
        return (round(min(x1, x2), 2), round(max(y1, y2), 2))

    def get_size(self):
        return (round(self.length, 2), round(self.width, 2))

    def move(self, dx, dy):
        self.corner1 = (self.corner1[0] + dx, self.corner1[1] + dy)
        self.corner2 = (self.corner2[0] + dx, self.corner2[1] + dy)

    def resize(self, width, height):
        x1, y1 = self.get_pos()
        self.corner1 = (x1, y1)
        self.corner2 = (x1 + width, y1 - height)
        self.length = abs(self.corner2[0] - self.corner1[0])
        self.width = abs(self.corner2[1] - self.corner1[1])

    def turn(self):
        center_x = (self.corner1[0] + self.corner2[0]) / 2
        center_y = (self.corner1[1] + self.corner2[1]) / 2
        
        self.corner1 = (center_x - self.width / 2, center_y - self.length / 2)
        self.corner2 = (center_x + self.width / 2, center_y + self.length / 2)
        
        self.length, self.width = self.width, self.length

    def scale(self, factor):
        center_x = (self.corner1[0] + self.corner2[0]) / 2
        center_y = (self.corner1[1] + self.corner2[1]) / 2
        
        half_length = self.length * factor / 2
        half_width = self.width * factor / 2
        
        self.corner1 = (center_x - half_length, center_y - half_width)
        self.corner2 = (center_x + half_length, center_y + half_width)
        
        self.length = abs(self.corner2[0] - self.corner1[0])
        self.width = abs(self.corner2[1] - self.corner1[1])
```

```python
# 8.  

class Cell:
    
    def __init__(self, state):
        self.state = state

    def status(self):
        return self.state


class Checkers:
    
    def __init__(self):
        self.board = {}
        columns = 'ABCDEFGH'
        rows = '12345678'

        for col in columns:
            for row in rows:
                pos = col + row
                if (int(row) + ord(col) - ord('A')) % 2 != 0:
                    if int(row) <= 3:
                        self.board[pos] = Cell('W')
                    elif int(row) >= 6:
                        self.board[pos] = Cell('B') 
                    else:
                        self.board[pos] = Cell('X') 
                else:
                    self.board[pos] = Cell('X')

    def move(self, f, t):
        self.board[t] = self.board[f]
        self.board[f] = Cell('X')

    def get_cell(self, p):
        return self.board[p]
```

```python
# 9.  

class Queue:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop(0)
        else:
            return None 

    def is_empty(self):
        return len(self.items) == 0
```

```python
# 10.  

class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        else:
            return None 

    def is_empty(self):
        return len(self.items) == 0
```
