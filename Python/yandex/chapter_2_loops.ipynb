```python
# 10.  

x, y = 0, 0
while (direction := input()) != "СТОП":
    steps = int(input())
    
    if direction == "СЕВЕР":
        x += steps
    elif direction == "ЮГ":
        x -= steps
    elif direction == "ВОСТОК":
        y += steps
    elif direction == "ЗАПАД":
        y -= steps
print(x, y, sep="\n")
```

```python
# 11.  

n = int(input())
sum_of_digits = 0
while n > 0:
    sum_of_digits += n % 10
    n //= 10
print(sum_of_digits)
```

```python
# 12.  


n = int(input())
max_digit = 0
while n > 0:
    current_digit = n % 10
    if current_digit > max_digit:
        max_digit = current_digit
    n //= 10
print(max_digit)
```

```python
# 13.  

n = int(input())
first_player = None
for _ in range(n):
    name = input().strip()
    if first_player is None or name < first_player:
        first_player = name
print(first_player)
```

```python
# 14.  

n = int(input())
if n < 2:
    print("NO")
else:
    is_prime = True
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            is_prime = False
            break
    if is_prime:
        print("YES")
    else:
        print("NO")
```

```python
# 15.  

N = int(input())
hare: int = 0
for _ in range(N):
    area = input()
    if "зайка" in area:
        hare += 1
print(hare)
```

```python
# 16.  

n = int(input())
original_str = str(n)
reversed_str = ""
i = len(original_str) - 1
while i >= 0:
    reversed_str += original_str[i]
    i -= 1
if original_str == reversed_str:
    print("YES")
else:
    print("NO")
```

```python
# 17.  

n = input()
result = ""
i = 0
while i < len(n):
    digit = int(n[i])
    if digit % 2 != 0:
        result += str(digit)
    i += 1
if result == "":
    result = "0"
print(result)
```

```python
# 18.  

n = int(input())
factors = []
divisor = 2
while n > 1:
    while n % divisor == 0:
        factors.append(divisor)
        n //= divisor
    divisor += 1
result = " * ".join(map(str, factors))
print(result)
```
