### Что такое Python  

Python - это бесплатный интерпретируемый высокоуровневый язык программирования общего назначения с открытым исходным кодом. Был создан Гвидо ван Россумом и выпущен в 1991 году. Он поддерживает несколько парадигм программирования, включая структурное (особенно процедурное), объектно-ориентированное и функциональное программирование. Python гибок в настройке и может быть расширен на другие платформы. Является одним из самых популярных языков программирования.

### Интерпретаторы Python  

Интерпретатор - это разновидность компьютерной программы, которая непосредственно выполняет инструкции, написанные на каком-либо языке программирования или языке сценариев. При этом предварительная компиляция кода не требуется.  

7 самых популярных интерпретаторов Python:  
1. IPython  
2. CPython (C + Python)  
3. IronPython    
4. Jython (Python + Java)  
5. PyPy (поддерживает серверные части C, CLI и JVM) - самый быстрый  
6. PythonNet  
7. Stackless Python  

### Версии Python  

1. Python 2  
Появились документы PEP,  сборщик мусора с обнаружением циклов для автоматизации управления памятью, расширенная поддержка 
Unicode для стандартизации символов и списочные включения.  

2. Python 2.7  
Вышел после Python 3 в 2010 году, планировался как последний выпуск в линейке версий 2.х. Целью было облегчить пользователям Python 2.х переход на Python 3 и предоставить совместимость между этими версиями. Добавлены новые модули: unittest, argparse, а также более удобные классы в коллекциях.  

3. Python 3
Выпущен в 2008 году после масштабной переработки. В нем были устранены и исправлены внутренние недостатки, присущие более ранним версиям языка. Основное внимание при разработке Python 3 уделялось очистке кодовой базы и устранению избыточности. К  основным изменениям в Python 3.0 можно отнести превращение оператора print во встроенную функцию, улучшение методики деления целых чисел и расширение поддержки Unicode.

### Приложения Anaconda  

3 основные программы, которые используются чаще всего:  
1. `Консоль IPython Qt` - это консоль, выполненная с помощью Qt, подходящая для работы с ядрами Jupyter, с поддержкой вывода мультимедиа, экспорта сеансов и многих других вещей. 
 - Консоль Qt - это очень легкое приложение, во многом похожее на командную 
строку, но с дополнительными функциями, которые можно реализовать только в графическом интерфейсе.  
 - Чтобы поближе познакомиться с этой консолью, наберите команду `?` - выведет краткое описание ее основных функций.  
 - В интерфейсе Qt для навигации по тексту используются привязки в стиле emacs. Их на данный момент настроить нельзя.  
 - Консоль IPython позволяет выполнять команды, а также вводить, взаимодействовать и визуализировать данные через любое количество полноценных интерпретаторов IPython. Каждая консоль выполняется в отдельном процессе, что позволяет запускать скрипты, прерывать выполнение и перезапускать или завершать сессию, не затрагивая другие консоли, и легко тестировать свой код в чистой среде, не прерывая основного сеанса.  
2. `Spyder IDE`  
 - Бесплатная интегрированная среда разработки (IDE), которая входит в состав Anaconda.  
 - Написана на Python.  
 - Имеет встроенную интеграцию со многими популярными научными пакетами, такими как NumPy, SciPy, Pandas, Matplotlib, SymPy и др.  
 - Основные компоненты: редактор текста, IPython консоль, обозреватель переменных (позволяет просматривать содержимое пространства имен и управлять им), справка.  
3. `Jupiter Notebook`  
Объединяет в себе 2 компонента:  
 - Веб-приложение. Основные возможности:  
    * Позволяет работать с кодом в редакторе, встроенном в окно браузера, с автоматическим выделением синтаксиса цветом, отступами, заполнением, табуляцией и анализом. Можно запустить код Python из браузера и получить прикрепленные к коду результаты вычислений. Отображение результатов вычислений выполняется с использованием мультимедийных форматов, таких как НТМL, LaTeX, PNG, SVG и т. д.  
 - Документы Notebook, содержащие все видимое в веб-браузере. Документы Notebook содержат входные и выходные данные интерактивного сеанса, а также дополнительный текст, который добавлен к коду, но не предназначен для выполнения. В файлах Notebook хранится полная запись сеанса работы в веб-приложении.  Эти документы являются внутренними файлами JSON и сохраняются с расширением `ipynb`.  Кроме того, любой документ Notebook формата ipynb, доступный по публичному URL-адресу, можно опубликовать через Jupyter Notebook Viewer (nbviewer). Этот сервис загружает документ Notebook по URL-адресу и отображает его как статическую веб-страницу. 

### Запуск сервера Notebook  

Вы можете запустить сервер Notebook из командной строки, используя следующую 
команду: 
```
Jupyter notebook 
```
В консоли появится некоторая информация о сервере, после чего в веб-браузере откроется URL-адрес веб-приложения (по умолчанию `http://127.0.0.1:8888`).  

При запуске сервера Notebook из командной строки вы также можете напрямую открыть конкретный документ, минуя дашборд, с помощью команды: 
```
Jupyter notebook my_notebook.ipynb
``` 

 - Если расширение файла не задано, по умолчанию добавляется расширение `ipynb`.  
 - Новый документ можно создать либо на панели управления, либо с помощью пункта меню File|New.  
 - Структура документа Notebook - последовательность ячеек (ячейки кода, markdown, неформатированный текст).  
 - Горячие клавиши для работы с документами:  
    * `Shift+Enter` -выполнить код в ячейке.  
    * `Esc` - командный режим.  
    * `Enter` - режим редактирования.  
    * `H` - командный режим ( получение списка всех доступных сочетаний клавиш).   

Одно из ключевых преимуществ документа Jupyter Notebook - возможность отображать графики, сгенерированные в результате выполнения ячеек кода. Ядро IPython специально спроектировано для бесшовной работы с библиотекой построения графиков Matplotlib, за счет которой и реализована эта функциональность. 2.12.1 Вопросы и ответы  

1. Python - это программное обеспечение с открытым исходным кодом. Это то же самое, что и бесплатное ПО?  

Программное обеспечение с открытым исходным кодом (Open Source) и бесплатное программное обеспечение (Free Software) - это два связанных, но не идентичных понятия:  
 * Открытый исходный код (OpenSource) - ПО, исходный код которого доступен для изучения, модификации и распространения. Цель: поддержка сотрудничества и развития с использованием общего исходного кода. 
 * Бесплатное программное обеспечение (Free Software) - ПО, которое можно использовать, копировать, изучать, изменять и распространять без ограничений. Цель: Гарантия четырех свобод пользователю: использования, изучения, распространения и улучшения программного обеспечения.  

Их общие черты: доступность и сообщество.  
Различия: Лицензии и философия.  

Python является примером программного обеспечения с открытым исходным кодом, которое также является бесплатным программным обеспечением.  

2. У всех ли бесплатных программ открытый исходный код? А если нет, то в чем разница?  

Нет, не у всех. Основное различие между бесплатным программным обеспечением и программным обеспечением с открытым исходным кодом заключается в доступности и свободе работы с исходным кодом. Бесплатное программное обеспечение можно использовать без оплаты, но оно не предоставляет возможности для изменения и распространения, в то время как программное обеспечение с открытым исходным кодом предоставляет эти свободы.  

3. Python поддерживает динамическую типизацию. Что это такое?  

Динамическая типизация в Python означает, что переменные в языке Python не имеют фиксированного типа данных. Вместо этого тип данных определяется автоматически во время выполнения программы, в зависимости от значения, которое присваивается переменной.

6. Python портативен. Что в этом контексте означает «портативность»?

Благодаря открытому исходному коду Python был перенесен на многие платформы, и его можно использовать в Windows, GNU/Linux, Macintosh, Solaris, Palm OS, PlayStation, Sharp Zaurus, Windows СЕ и PocketPC! 

7. В чем разница между «расширяемым» и «встраиваемым» языком?  

Расширяемый язык программирования позволяет разработчикам добавлять новые функции, библиотеки и возможности без изменения самого ядра языка.  

Встраиваемый язык программирования предназначен для интеграции (встраивания) в другие приложения и системы для выполнения скриптов и автоматизации задач.  

8. В чем смысл IDE? Чем она отличается от командной строки?  

    * IDE представляет более широкие возможности, включая редактор кода, отладчик, компилятор и другие инструменты в одном месте, тогда как командная строка фокусируется на выполнении команд.  
    * IDE имеет графический интерфейс, в то время как командная строка является текстовым интерфейсом.  
    * IDE повышает продуктивность разработчиков благодаря интеграции всех необходимых инструментов, тогда как командная строка требует от разработчика выполнения команд вручную.  

9. Как открыть существующий документ Jupyter Notebook? Чем эта процедура отличается от открытия РDF-файла или текстового файла?  

Каждому открытом документу соответствует один интерактивный сеанс, подключенный к ядру, которое будет выполнять пользовательский код и возвращать результаты. Это ядро остается активным при закрытии окна браузера, и повторное открытие того же документа с панели управления приведет к повторному подключению веб-приложения к тому же ядру. На панели управления у документов с активным ядром отображается кнопка Shutdown, а у документов без активного ядра вместо нее отображается кнопка Delete.  

Отличия от открытия РDF-файла или текстового файла: 
* Jupyter Notebook: Открывается в веб-браузере через веб-интерфейс, поддерживает выполнение кода и отображение интерактивных результатов.  
* PDF-файл: Открывается в программах для чтения PDF с возможностью просмотра и аннотирования.  
* Текстовый файл: Открывается в текстовых редакторах для редактирования простого текста.  

10. В чем разница между «ячейками разметки Markdown» и «ячейками кода» в Jupyter Notebook? Для чего они нужны?  

Ячейки кода можно запускать. Ячейки markdown не будут запускаться. Они предназначены для написания текстовых описаний, документации, заголовков и форматированного текста.
